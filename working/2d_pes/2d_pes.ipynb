```python
import sys
import numpy as np
import jax.numpy as jnp
from matplotlib import pyplot as plt
from plotly import graph_objects as go

sys.path.append("src")
from pypes import MB_PES2D, LEPS_PES2D

# import plotly.io as pio
# pio.renderers.default = "browser"
```

```python
pes = MB_PES2D(backend="auto")
x = np.linspace(pes.x_range[0], pes.x_range[1], pes.resolution)
y = np.linspace(pes.y_range[0], pes.y_range[1], pes.resolution)

X, Y = np.meshgrid(x, y)
Z = pes.energy([X, Y])
Z_mask = np.where(Z > pes.v_max, pes.v_max, Z)

fig = go.Figure(go.Surface(x=X, y=Y, z=Z_mask, colorscale="plasma",
        opacity=0.8,
        contours={
            "z": {
                "show": True,
                "usecolormap": False,
                "highlightcolor": "black",
                "project_z": False,
                "start": Z.min().item(),
                "end": pes.v_max,
                "size": 3,
                "color": "white",
                "width": 2,
            }},
        colorbar=dict(
            title=dict(
                text="Potential Energy (kJ/mol)",
                font=dict(size=14)
            ),
            thickness=20
        )
        ),
        )

fig.update_layout(
    template="plotly_dark",
    title=dict(text='Müller-Brown Potential'),
    autosize=False,
    scene=dict(
        xaxis = dict(
            gridcolor="white"
        ),
        yaxis = dict(
            gridcolor="white"
        ),
        zaxis = dict(
            gridcolor="white"
        ),
        xaxis_title="X coordinate",
        yaxis_title="Y coordinate",
        zaxis_title="Potential Energy (kJ/mol)",
        bgcolor="black",
        ),
    width=700,
    height=500,
    )

fig.write_html("pes_mb.html", include_plotlyjs="cdn")
# fig.show()
```
