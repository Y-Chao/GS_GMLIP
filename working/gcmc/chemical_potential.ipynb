## Convert a formula to symbol list

```python
def formula_to_symbols(formula: str) -> list:
    from ase.data import atomic_numbers, chemical_symbols
    import re

    pattern = r'([A-Z][a-z]?)(\d*)'
    matches = re.findall(pattern, formula)

    symbols = []
    for (element, count) in matches:
        if element not in chemical_symbols:
            raise ValueError(f"Element '{element}' is not recognized.")
        count = int(count) if count else 1
        symbols.extend([element] * count)

    return symbols

formula_to_symbols("H2Oh")  # Example usage
```
